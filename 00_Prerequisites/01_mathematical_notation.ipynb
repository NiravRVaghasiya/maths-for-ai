## 📐 Theory

Every field develops shorthand for ideas it uses constantly. Mathematics is no exception —
and machine learning papers lean on that shorthand *hard*. This notebook is a dictionary:
by the end, symbols that currently look like alphabet soup will read like plain English.

### Summation ($\sum$) and Product ($\prod$) notation

Summation notation is just a compact "add up all of these" instruction:

$$\sum_{i=1}^{n} x_i = x_1 + x_2 + \cdots + x_n$$

Read this as: "start a running total at $i=1$, add $x_i$, increment $i$, repeat until $i=n$."
The letter under $\sum$ ($i=1$) is the **starting index**, the number on top ($n$) is the
**stopping point**, and $x_i$ is the **term being summed**.

Product notation ($\prod$) is the multiplicative twin:

$$\prod_{i=1}^{n} x_i = x_1 \times x_2 \times \cdots \times x_n$$

Nested sums are common once you have more than one index to loop over — for example, summing
every entry of an $m \times n$ matrix $A$:

$$\sum_{i=1}^{m}\sum_{j=1}^{n} A_{ij}$$

This is exactly two nested `for` loops: the outer sum walks down rows, the inner sum walks
across columns of each row.

### Set-builder notation

Set-builder notation describes a set by a *rule* rather than by listing every member:

$$\{x \in \mathbb{R} \mid x > 0\}$$

Read the vertical bar $\mid$ as "such that." This says: "the set of all real numbers $x$
such that $x$ is positive." The symbol $\mathbb{R}$ denotes the real numbers; you'll also
see $\mathbb{N}$ (natural numbers $0,1,2,\dots$), $\mathbb{Z}$ (integers), and $\mathbb{R}^n$
($n$-dimensional real vectors — the space every embedding and weight matrix lives in).

### Logical quantifiers

Two symbols let you make precise claims about "all" or "some" members of a set:

- $\forall$ ("for all") — $\forall x \in S,\ P(x)$ means every element of $S$ satisfies property $P$.
- $\exists$ ("there exists") — $\exists x \in S,\ P(x)$ means at least one element of $S$ satisfies $P$.

For example, "a function $f$ is bounded" can be written precisely as
$\exists M \in \mathbb{R} \text{ such that } \forall x,\ |f(x)| \le M$ — there is *some* bound $M$
that works for *every* input.

### Index and tensor notation

Subscripts pick out individual entries of a vector, matrix, or higher-dimensional array:

- $x_i$ — the $i$-th entry of vector $x$.
- $A_{ij}$ — the entry in row $i$, column $j$ of matrix $A$.
- $T_{ijk}$ — the entry at position $(i,j,k)$ of a rank-3 tensor $T$.

A special convention called the **Einstein summation convention** drops the $\sum$ symbol
whenever an index is repeated: $A_{ij} B_{jk}$ implicitly means $\sum_j A_{ij} B_{jk}$, which
is exactly matrix multiplication. You'll meet this again — with a vengeance — in
[Notebook 01.09](../01_Linear_Algebra/09_tensor_operations.ipynb) when we cover `einsum`.

## 👁️ Visual Intuition

A sum is just an accumulation process. Watching the *running total* of
$\sum_{i=1}^{n} i$ climb bar by bar makes the abstract $\sum$ symbol concrete: it's nothing
more than a loop with a running total, drawn out over time.

```python
# Visualize summation as an accumulation process
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')

n = 10
terms = np.arange(1, n + 1)          # x_i = i
running_total = np.cumsum(terms)     # the partial sums \sum_{i=1}^{k} i, for k = 1..n

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(terms, terms, color="#4C72B0")
axes[0].set_title(r"The terms being summed: $x_i = i$")
axes[0].set_xlabel("$i$")
axes[0].set_ylabel("$x_i$")

axes[1].plot(terms, running_total, marker="o", color="#DD8452")
axes[1].set_title(r"Running total: $\sum_{i=1}^{k} i$ as $k$ grows")
axes[1].set_xlabel("$k$")
axes[1].set_ylabel("Partial sum")

plt.tight_layout()
plt.show()

print(f"Final value of sum_{{i=1}}^{{{n}}} i =", running_total[-1],
      "  (matches the closed form n(n+1)/2 =", n * (n + 1) // 2, ")")
```

## 🐍 Implementation from Scratch

Let's translate summation, product, and quantifier notation directly into Python, so the
symbols stop being abstract and start being *executable*.

```python
# Sigma / Pi notation as literal Python translations
import numpy as np

def sigma(terms):
    # Sigma: sum a sequence of terms -- the direct translation of \sum_i x_i
    total = 0
    for x in terms:
        total += x
    return total

def pi(terms):
    # Pi: multiply a sequence of terms -- the direct translation of \prod_i x_i
    product = 1
    for x in terms:
        product *= x
    return product

def for_all(collection, predicate):
    # \forall x in collection, predicate(x) -- True only if EVERY element passes
    return all(predicate(x) for x in collection)

def there_exists(collection, predicate):
    # \exists x in collection, predicate(x) -- True if AT LEAST ONE element passes
    return any(predicate(x) for x in collection)

xs = [1, 2, 3, 4, 5]
print("sum_{i=1}^{5} x_i         =", sigma(xs))
print("numpy check (np.sum)      =", np.sum(xs))
print("prod_{i=1}^{5} x_i        =", pi(xs))
print("numpy check (np.prod)     =", np.prod(xs))
print("forall x in xs, x > 0     =", for_all(xs, lambda x: x > 0))
print("exists x in xs, x > 4     =", there_exists(xs, lambda x: x > 4))

# Nested summation: sum_i sum_j A_{ij} over a matrix
A = np.array([[1, 2, 3], [4, 5, 6]])
nested_sum = sigma(sigma(row) for row in A)
print("\nNested sum_i sum_j A_ij   =", nested_sum)
print("numpy check (A.sum())     =", A.sum())
```

## 🤖 Why This Matters for AI

Open any ML paper and the *math* is rarely the hard part — the **notation** is what trips
people up first. Softmax, cross-entropy, and attention are all defined with exactly the
symbols above:

$$\text{softmax}(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}} \qquad
\mathcal{L} = -\sum_{i=1}^{N} y_i \log(\hat{y}_i)$$

The softmax denominator is a $\sum$ over every class score; the cross-entropy loss is a
$\sum$ over every training example. If you can read the sigma, you can read the loss
function. Below, we implement both directly from their notation and confirm they match
library implementations.

```python
# Reading loss-function notation literally: softmax and cross-entropy from scratch
import numpy as np

def softmax(z):
    # softmax(z)_i = exp(z_i) / sum_j exp(z_j) -- read straight off the formula
    z = z - np.max(z)  # numerical stability trick, covered in Notebook 06.02
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z)   # the sum_{j=1}^{K} in the denominator

def cross_entropy(y_true_onehot, y_pred_probs, eps=1e-12):
    # L = -sum_i y_i * log(yhat_i) -- one term per class, only the true class survives.
    # Clip probabilities away from exactly 0 before taking log: if y_pred_probs contains a
    # true zero (a saturated softmax, or hard 0/1 probabilities passed in directly), log(0)
    # is -inf, and -inf * 0 (the one-hot entries for every OTHER class) evaluates to nan in
    # IEEE754 -- silently poisoning the whole sum even though those terms should contribute
    # exactly 0. Clipping to [eps, 1] keeps every log() argument strictly positive.
    y_pred_probs = np.clip(y_pred_probs, eps, 1.0)
    return -np.sum(y_true_onehot * np.log(y_pred_probs))

logits = np.array([2.0, 1.0, 0.1])       # raw model scores for 3 classes
probs = softmax(logits)
print("logits          :", logits)
print("softmax(logits) :", np.round(probs, 4), "  (sums to", round(probs.sum(), 4), ")")

y_true = np.array([1, 0, 0])              # one-hot label: class 0 is correct
loss = cross_entropy(y_true, probs)
print("cross-entropy loss:", round(loss, 4))
print("equivalent to -log(p_true):", round(-np.log(probs[0]), 4))

# --- Regression test: a probability vector containing an exact 0 (e.g. an extremely
# confident, saturated prediction) must NOT poison the loss with nan. ---
probs_saturated = np.array([1.0, 0.0, 0.0])   # a model that is (numerically) 100% confident
y_true_other = np.array([0, 1, 0])             # but the TRUE class is one it assigned exactly 0 to
loss_saturated = cross_entropy(y_true_other, probs_saturated)
print("\n--- Regression test: probability vector with an exact 0.0 entry ---")
print("probs:", probs_saturated, " true class: index 1 (assigned probability exactly 0)")
print(f"cross-entropy loss: {loss_saturated:.4f}  (finite and large, as it should be for a")
print(f"confidently WRONG prediction -- not nan, which the unclipped log(0)*0 would produce)")
assert np.isfinite(loss_saturated), "cross_entropy must stay finite even with a true-zero probability"
```

### 💭 UNDERSTAND
2. $\sum_{i=1}^{n}\sum_{j=1}^{m} A_{ij}$ and $\sum_{j=1}^{m}\sum_{i=1}^{n} A_{ij}$ (swapping
   which index's sum is on the outside) always give the same value for any real matrix $A$.
   Explain *why* in one sentence — what property of addition makes the order of the two nested
   sums irrelevant here — and give one operation that does NOT have this order-independence
   property (so that swapping a nested loop's order for it would change the answer).

<details>
<summary>💡 Solution</summary>

Addition is commutative and associative, so any finite sum of real numbers gives the same total
regardless of the order the terms are added in — nested $\sum\sum$ is just "add up every
$A_{ij}$," and *which* index you loop over first doesn't change that it's still the same
multiset of numbers being added. Matrix multiplication is a natural non-example: $AB \ne BA$ in
general, so "sum over $k$ of $A_{ik}B_{kj}$" is not interchangeable with swapping the roles of
$A$ and $B$ the way the two orders of a plain nested sum are.

</details>

### ✏️ DERIVE
3. Using only the definition $\sum_{i=1}^{n} x_i = x_1 + \cdots + x_n$ and ordinary algebra
   (no induction needed), show that $\sum_{i=1}^{n}(a x_i + b) = a\sum_{i=1}^{n} x_i + nb$ for
   constants $a, b$. Then use this identity to explain, in symbols, why
   $\frac{1}{n}\sum_{i=1}^n (x_i - c)$ equals $\bar{x} - c$ for any constant $c$, where
   $\bar{x} = \frac{1}{n}\sum_i x_i$ is the mean.

<details>
<summary>💡 Solution outline</summary>

Expand the left side term by term: $\sum_{i=1}^n (ax_i + b) = (ax_1+b) + (ax_2+b) + \cdots +
(ax_n+b)$. Regroup (associativity/commutativity of addition) into the $ax_i$ terms and the $b$
terms separately: $= (ax_1 + ax_2 + \cdots + ax_n) + (b + b + \cdots + b) = a(x_1+\cdots+x_n) +
nb = a\sum_i x_i + nb$, exactly the claim.

For the second part, set $a=1$ (so $ax_i = x_i$) and $b = -c$: $\sum_i (x_i - c) = \sum_i x_i -
nc$. Dividing both sides by $n$: $\frac{1}{n}\sum_i(x_i-c) = \frac{1}{n}\sum_i x_i - c =
\bar{x} - c$. This is exactly why "center the data by subtracting its mean" and "the mean of
the centered data is 0" are the same fact — set $c = \bar{x}$ and the right side becomes
$\bar{x} - \bar{x} = 0$.

</details>

### 🤖 APPLY
5. The mean squared error loss is $\text{MSE} = \frac{1}{n}\sum_{i=1}^n (y_i -

<details>
<summary>✅ What you should observe</summary>

Your `sigma`-based implementation and `np.mean((y - yhat)**2)` should agree (up to floating-point

### 🚀 CHALLENGE
6. The KL divergence between two discrete distributions $p,q$ over the same $K$ outcomes is
   $D_{KL}(p\|q) = \sum_{i=1}^K p_i \log(p_i/q_i)$ (you'll meet this properly in `05.02`).
   Using only `sigma()`, `np.log`, and the fact that a valid PMF satisfies $\sum_i p_i = 1$,
   $p_i \ge 0$, implement it, and check numerically that it's zero when $p=q$ and strictly
   positive for at least three different $p\ne q$ pairs you construct yourself (fix a small
   numerical floor like `1e-12` inside the log to avoid `log(0)` if any entry is exactly 0).

<details>
<summary>🔍 What a strong answer covers</summary>

A correct implementation reads the formula off directly: `sigma(p[i] * np.log(p[i]/q[i]) for i in
range(K))`. A strong answer notices and states *why* $D_{KL}(p\|p)=0$ always (every term
becomes $p_i\log(1) = 0$) without needing to run anything, then confirms it numerically as a
sanity check rather than the other way around, and separately confirms strict positivity for
several genuinely different $p \ne q$ pairs (not just one) since a single example could
coincidentally land near zero. The strongest answers also note that this sum is finite as
written but conceptually different from summing over a continuous outcome space, foreshadowing
why KL divergence for continuous distributions (`05.02`) needs an integral instead of a sum.
