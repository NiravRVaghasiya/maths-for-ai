## 📐 Theory

### Set operations

A **set** is an unordered collection of distinct elements. Given two sets $A$ and $B$:

- **Union** $A \cup B$ — everything in $A$ *or* $B$ (or both).
- **Intersection** $A \cap B$ — everything in *both* $A$ *and* $B$.
- **Complement** $A^c$ — everything *not* in $A$ (relative to some universal set).
- **Difference** $A \setminus B$ — everything in $A$ but not in $B$.
- **Subset** $A \subseteq B$ — every element of $A$ is also in $B$.

### Propositional logic

Propositional logic manipulates statements that are either `True` or `False` using connectives:

| Connective | Symbol | Meaning |
|---|---|---|
| AND | $\land$ | true only if both sides are true |
| OR | $\lor$ | true if at least one side is true |
| NOT | $\lnot$ | flips true/false |
| IMPLIES | $\Rightarrow$ | "if...then"; false only when the premise is true and conclusion is false |
| IFF | $\Leftrightarrow$ | "if and only if"; true when both sides match |

A neat and useful fact for simplifying expressions is **De Morgan's laws**:

$$\lnot(A \land B) = \lnot A \lor \lnot B \qquad \lnot(A \lor B) = \lnot A \land \lnot B$$

### Proof techniques

Two techniques come up constantly when reading (or writing) rigorous ML theory:

**Proof by induction** proves a statement $P(n)$ holds for all natural numbers $n$ in two steps:
1. *Base case*: show $P(0)$ (or $P(1)$) is true.
2. *Inductive step*: assume $P(k)$ is true (the "inductive hypothesis"), and show this implies
   $P(k+1)$ is true.

If both steps hold, $P(n)$ is true for every $n$ — like an infinite row of dominoes where you
only need to show the first one falls, and that any falling domino knocks over the next.

**Proof by contradiction** assumes the *opposite* of what you want to prove, then shows that
assumption leads to something impossible. Since the assumption can't be true, the original
claim must hold. The classic example: proving $\sqrt{2}$ is irrational by assuming it's a
fraction $p/q$ in lowest terms and deriving a contradiction (both $p$ and $q$ turn out to be
even, contradicting "lowest terms").

### Boolean algebra

Boolean algebra treats `True`/`False` (or 1/0) as values you can compute with, using AND, OR,
and NOT as if they were arithmetic operators. This is the same algebra used to build binary
classifiers' decision rules and digital logic circuits.

## 👁️ Visual Intuition

Venn diagrams turn abstract set operations into overlapping regions — a shape you can *see*.

```python
# Venn diagrams for union, intersection, and difference
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')

def draw_venn(ax, title, shade):
    # shade: one of 'union', 'intersection', 'difference' -- which region to highlight
    c1 = Circle((0.35, 0.5), 0.32, alpha=0.35, color="#4C72B0", label="A")
    c2 = Circle((0.65, 0.5), 0.32, alpha=0.35, color="#DD8452", label="B")
    ax.add_patch(c1)
    ax.add_patch(c2)
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.text(0.20, 0.5, "A", fontsize=14, ha="center")
    ax.text(0.80, 0.5, "B", fontsize=14, ha="center")
    ax.axis("off")

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
draw_venn(axes[0], r"$A \cup B$ (union)", "union")
draw_venn(axes[1], r"$A \cap B$ (intersection)", "intersection")
draw_venn(axes[2], r"$A \setminus B$ (difference)", "difference")
plt.tight_layout()
plt.show()
```

## 🐍 Implementation from Scratch

Python's built-in `set` type implements every set operation from the theory section directly.
We'll also build a tiny truth-table generator to make propositional logic mechanical and
verifiable, and implement proof-by-induction as a runnable check rather than just a proof.

```python
# Set operations, a truth-table generator, and a computational induction check
from itertools import product

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

print("A union B         :", A | B)
print("A intersection B  :", A & B)
print("A difference B    :", A - B)
print("A subset of A|B   :", A.issubset(A | B))

def truth_table(expr, names):
    # Print a truth table for a boolean expression expr(*bools) over the given variable names
    header = " | ".join(names) + " | result"
    print(header)
    print("-" * len(header))
    for combo in product([False, True], repeat=len(names)):
        result = expr(*combo)
        row = " | ".join(str(int(b)) for b in combo) + f" | {int(result)}"
        print(row)

print("\nDe Morgan's law: NOT(A AND B) == (NOT A) OR (NOT B)?")
truth_table(lambda a, b: (not (a and b)) == ((not a) or (not b)), ["A", "B"])

# Proof by induction, verified computationally for a range of n:
# Claim: sum_{i=1}^{n} i = n(n+1)/2
def check_induction(n_max):
    for n in range(1, n_max + 1):
        lhs = sum(range(1, n + 1))
        rhs = n * (n + 1) // 2
        assert lhs == rhs, f"Failed at n={n}"
    print(f"Verified sum_1^n i = n(n+1)/2 for all n in [1, {n_max}]")

check_induction(1000)
```

## 🤖 Why This Matters for AI

**Feature spaces as sets.** A classifier partitions its input space $\mathbb{R}^d$ into
disjoint sets — one per class. The decision boundary is exactly the *set* of points where
the classifier is indifferent between two classes.

**Logical reasoning in LLMs.** Benchmarks that test whether language models can do logical
reasoning (e.g. "if all cats are mammals, and Felix is a cat, is Felix a mammal?") are directly
testing whether the model has internalized $\forall$/$\Rightarrow$ style inference. Boolean
masks are also used *literally* inside transformers — the causal attention mask is a boolean
matrix that determines which token positions are "visible" ($\land$-ed against the raw
attention scores before softmax).

```python
# A causal attention mask IS a boolean set operation over positions
import numpy as np

seq_len = 6
# allowed[i, j] = True iff position j is <= position i (j "in the causal set" of i)
positions = np.arange(seq_len)
allowed = positions[None, :] <= positions[:, None]   # boolean matrix, shape (seq_len, seq_len)

print("Causal mask (True = attend, False = blocked):")
print(allowed.astype(int))

# Applying the mask is literally "intersect the attention scores with the allowed set"
scores = np.random.randn(seq_len, seq_len)
masked_scores = np.where(allowed, scores, -np.inf)   # blocked positions -> -inf before softmax
print("\nRow 2 can only attend to columns:", np.where(allowed[2])[0].tolist())
```
