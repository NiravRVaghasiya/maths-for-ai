## 📐 Theory

### Shattering and the growth function

Let $\mathcal{H}$ be a hypothesis class of binary classifiers $h : \mathcal{X} \to \{0,1\}$.
Given a finite set $C = \{x_1, \ldots, x_m\} \subset \mathcal{X}$, the **restriction** of
$\mathcal{H}$ to $C$ is the set of all labeling patterns achievable by some $h \in \mathcal{H}$:

$$\mathcal{H}|_C \;=\; \bigl\{\bigl(h(x_1),\ldots,h(x_m)\bigr) : h \in \mathcal{H}\bigr\} \;\subseteq\; \{0,1\}^m$$

We say $\mathcal{H}$ **shatters** $C$ if $|\mathcal{H}|_C| = 2^m$ — that is, for *every* one of
the $2^m$ possible binary labelings of the $m$ points, there exists some $h \in \mathcal{H}$
that produces exactly that labeling.

The **growth function** (also called the **shattering coefficient**) measures the maximum number
of distinct labelings achievable on *any* set of $m$ points:

$$\Pi_{\mathcal{H}}(m) \;=\; \max_{C \subset \mathcal{X},\;|C|=m} |\mathcal{H}|_C|$$

Clearly $\Pi_{\mathcal{H}}(m) \le 2^m$ always. Shattering some set of size $m$ means
$\Pi_{\mathcal{H}}(m) = 2^m$.

### VC dimension

The **Vapnik–Chervonenkis (VC) dimension** of $\mathcal{H}$ is the largest $m$ such that
$\mathcal{H}$ shatters *some* set of size $m$:

$$\text{VCdim}(\mathcal{H}) \;=\; \max\{m : \Pi_{\mathcal{H}}(m) = 2^m\}$$

If $\mathcal{H}$ can shatter arbitrarily large sets, $\text{VCdim}(\mathcal{H}) = \infty$.
Two critical points: (1) to establish $\text{VCdim} \ge d$, you must exhibit *one specific*
set of $d$ points that $\mathcal{H}$ shatters; (2) to establish $\text{VCdim} < d+1$, you
must show that *no* set of $d+1$ points can be shattered — a universal quantifier over all
possible $(d+1)$-point configurations.

### Worked example 1: Threshold classifiers ($\text{VCdim} = 1$)

Consider $\mathcal{X} = \mathbb{R}$ and $\mathcal{H} = \{h_\theta : h_\theta(x) =
\mathbf{1}[x \ge \theta],\; \theta \in \mathbb{R}\}$. A single point $\{x_1\}$ is shattered:
choose $\theta > x_1$ for label $0$, $\theta \le x_1$ for label $1$. But no two-point set
$\{x_1, x_2\}$ with $x_1 < x_2$ can achieve the labeling $(1, 0)$ — if $x_1 \ge \theta$ then
certainly $x_2 \ge \theta$. So $\text{VCdim}(\mathcal{H}) = 1$.

### Worked example 2: Linear classifiers in $\mathbb{R}^d$ ($\text{VCdim} = d+1$)

Let $\mathcal{H}$ be the set of all homogeneous linear classifiers on $\mathbb{R}^{d+1}$
(equivalently, affine classifiers on $\mathbb{R}^d$ after appending a bias coordinate):
$h_w(x) = \mathbf{1}[w^\top x \ge 0]$, $w \in \mathbb{R}^{d+1}$.

**Lower bound ($\text{VCdim} \ge d+1$):** Take the $d+1$ standard basis vectors
$\{e_1, \ldots, e_{d+1}\}$ in $\mathbb{R}^{d+1}$. For any labeling
$(y_1,\ldots,y_{d+1}) \in \{0,1\}^{d+1}$, set $w_i = 2y_i - 1 \in \{-1,+1\}$. Then
$w^\top e_j = w_j = 2y_j - 1$, which is $\ge 0$ iff $y_j = 1$, producing exactly the desired
labeling. So the standard basis is shattered.

**Upper bound ($\text{VCdim} \le d+1$):** Take any $d+2$ points
$x_1,\ldots,x_{d+2} \in \mathbb{R}^{d+1}$. Since $d+2 > d+1 = \dim(\mathbb{R}^{d+1})$,
these points are linearly dependent (recall `01.03`): there exist coefficients
$\alpha_1,\ldots,\alpha_{d+2}$ (not all zero) with $\sum_i \alpha_i x_i = 0$. Define
$I^+ = \{i : \alpha_i > 0\}$ and $I^- = \{i : \alpha_i \le 0\}$. Consider the labeling
$y_i = \mathbf{1}[i \in I^+]$. Suppose for contradiction that some $w$ achieves this labeling:
$w^\top x_i \ge 0$ for $i \in I^+$ and $w^\top x_i < 0$ for $i \in I^-$. Then
$0 = w^\top(\sum_i \alpha_i x_i) = \sum_{i \in I^+} \alpha_i(w^\top x_i) +
\sum_{i \in I^-} \alpha_i(w^\top x_i)$. The first sum is $\ge 0$ (positive $\alpha_i$ times
non-negative $w^\top x_i$) and the second is $> 0$ (non-positive $\alpha_i$ times negative
$w^\top x_i$), so their sum is $> 0$ — a contradiction. No set of $d+2$ points can be
shattered, so $\text{VCdim} \le d+1$. Combined: $\text{VCdim} = d+1$. $\blacksquare$

### Worked example 3: Sinusoidal classifiers ($\text{VCdim} = \infty$) — a cautionary tale

Consider $\mathcal{X} = \mathbb{R}$ and $\mathcal{H} = \{h_\omega : h_\omega(x) =
\mathbf{1}[\sin(\omega x) \ge 0],\; \omega > 0\}$. For any $m$ points, one can choose
$\omega$ large enough that the sinusoid oscillates rapidly, achieving any desired labeling
pattern. Specifically, for $m$ points $x_1 < \cdots < x_m$, the frequency $\omega$ can be
tuned so that each $x_i$ falls in either a positive or negative half-period as desired. This
gives $\text{VCdim} = \infty$, despite $\mathcal{H}$ being parameterized by a single real
number — a stark warning that **the number of parameters does not determine VC dimension**.
A single-parameter class can have infinite VC dimension, while a $(d+1)$-parameter linear
classifier class has finite VC dimension $d+1$.

### Sauer–Shelah lemma

The key structural result bounding the growth function:

**Lemma (Sauer–Shelah).** *If $\text{VCdim}(\mathcal{H}) = d$, then for all $m \ge 1$:*

$$\Pi_{\mathcal{H}}(m) \;\le\; \sum_{i=0}^{d} \binom{m}{i}$$

*In particular, for $m \ge d$:*

$$\Pi_{\mathcal{H}}(m) \;\le\; \left(\frac{em}{d}\right)^d$$

The critical consequence: once $m$ exceeds $d$, the growth function transitions from
potentially exponential ($2^m$) to **polynomial** ($O(m^d)$). This phase transition is the
engine of all VC-based generalization bounds.

**Proof by induction on $m + d$.**

Define $\Phi(m, d) = \sum_{i=0}^{d} \binom{m}{i}$. We need $|\mathcal{H}|_C| \le \Phi(m, d)$
for any $C$ with $|C| = m$.

*Base cases:* If $d = 0$, then $\mathcal{H}$ shatters no single point, so $\mathcal{H}|_C$
contains at most one labeling: $|\mathcal{H}|_C| \le 1 = \Phi(m, 0)$. If $m = 0$, trivially
$|\mathcal{H}|_C| \le 1 = \Phi(0, d)$.

*Inductive step:* Fix an arbitrary $x_1 \in C$ and let $C' = C \setminus \{x_1\}$ with
$|C'| = m-1$. Partition $\mathcal{H}|_C$ into labelings where $x_1$ receives a fixed value:
let $\mathcal{H}_0 = \{(h(x_2),\ldots,h(x_m)) : h \in \mathcal{H}\}$ (the projection onto
$C'$, ignoring duplicates), and let $\mathcal{H}^\pm$ be the set of labelings on $C'$ that
appear with *both* values of $h(x_1)$. Then:

$$|\mathcal{H}|_C| \;=\; |\mathcal{H}_0| + |\mathcal{H}^\pm|$$

Now: $\mathcal{H}_0$ is a restriction of $\mathcal{H}$ to $m-1$ points, so by induction
$|\mathcal{H}_0| \le \Phi(m-1, d)$. And $\mathcal{H}^\pm$ shatters any set that
$\mathcal{H}$ shatters within $C'$ *plus* the point $x_1$ — so $\text{VCdim}(\mathcal{H}^\pm)
\le d-1$, giving $|\mathcal{H}^\pm| \le \Phi(m-1, d-1)$ by induction. Using Pascal's
identity $\Phi(m-1, d) + \Phi(m-1, d-1) = \Phi(m, d)$:

$$|\mathcal{H}|_C| \le \Phi(m-1, d) + \Phi(m-1, d-1) = \Phi(m, d) \quad\blacksquare$$

The polynomial bound $(em/d)^d$ follows from the standard combinatorial inequality
$\sum_{i=0}^{d} \binom{m}{i} \le (em/d)^d$ for $m \ge d \ge 1$.

### VC generalization bound

The growth function, bounded by Sauer–Shelah, enters the generalization bound via a
**symmetrization** argument (due to Vapnik and Chervonenkis). The key idea: to bound
$\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)|$ where $R(h) = \mathbb{E}[\ell(h(X), Y)]$
is the true risk and $\hat{R}_n(h) = \frac{1}{n}\sum_{i=1}^n \ell(h(X_i), Y_i)$ is the
empirical risk, we introduce a **ghost sample** $S' = \{(X'_1,Y'_1),\ldots,(X'_n,Y'_n)\}$
drawn independently from the same distribution. Since $R(h) \approx \hat{R}'_n(h)$ for the
ghost sample, bounding $\sup_h |\hat{R}_n(h) - R(h)|$ reduces to bounding
$\sup_h |\hat{R}_n(h) - \hat{R}'_n(h)|$.

Conditionally on the $2n$ points (both samples), there are at most
$\Pi_{\mathcal{H}}(2n)$ distinct label patterns — so the union bound over effectively
$\Pi_{\mathcal{H}}(2n)$ hypotheses (rather than the possibly infinite $|\mathcal{H}|$)
combined with Hoeffding's inequality (recall `09.04`) yields:

$$P\!\left(\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)| > \varepsilon\right)
\;\le\; 4\,\Pi_{\mathcal{H}}(2n)\,e^{-n\varepsilon^2/8}$$

Applying Sauer–Shelah ($\Pi_{\mathcal{H}}(2n) \le (2en/d)^d$) and solving for $\varepsilon$
at confidence $1 - \delta$:

$$\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)| \;\le\;
\sqrt{\frac{8}{n}\left(d\ln\frac{2en}{d} + \ln\frac{4}{\delta}\right)}$$

with probability at least $1 - \delta$. Equivalently, an $O\!\left(\sqrt{\frac{d \log(n/d)}{n}}\right)$ bound on the uniform generalization gap.

### Uniform convergence and ERM consistency

A hypothesis class $\mathcal{H}$ has the **uniform convergence property** if

$$\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)| \;\xrightarrow{\text{a.s.}}\; 0
\quad\text{as } n \to \infty$$

When this holds, **Empirical Risk Minimization (ERM)** is consistent: let $\hat{h}_n =
\arg\min_{h \in \mathcal{H}} \hat{R}_n(h)$ and $h^* = \arg\min_{h \in \mathcal{H}} R(h)$.
Then:

$$R(\hat{h}_n) - R(h^*) \;\le\; 2\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)|
\;\xrightarrow{\text{a.s.}}\; 0$$

The proof: $R(\hat{h}_n) \le \hat{R}_n(\hat{h}_n) + \sup|R - \hat{R}_n| \le
\hat{R}_n(h^*) + \sup|R - \hat{R}_n| \le R(h^*) + 2\sup|R - \hat{R}_n|$.
The VC bound guarantees uniform convergence whenever $d = \text{VCdim}(\mathcal{H}) < \infty$,
since the bound $\to 0$ as $n \to \infty$.

### Fundamental Theorem of Statistical Learning

For binary classification with 0-1 loss, the following are **equivalent**:

1. $\mathcal{H}$ is **PAC learnable** (there exists an algorithm that, given $\varepsilon, \delta > 0$, uses $n(\varepsilon, \delta)$ samples to find $h$ with $R(h) \le \min_{h' \in \mathcal{H}} R(h') + \varepsilon$ w.p. $\ge 1-\delta$).
2. $\mathcal{H}$ has **finite VC dimension**.
3. $\mathcal{H}$ has the **uniform convergence property**.

The implication $2 \Rightarrow 3$ is the VC bound derived above. $3 \Rightarrow 1$ follows
because ERM is consistent under uniform convergence. $1 \Rightarrow 2$ (the converse — infinite
VC dimension implies non-learnability) uses a probabilistic argument: if $\text{VCdim} = \infty$,
for any sample size $n$ one can construct a distribution where ERM (and in fact *any* learner)
fails with constant probability — the celebrated **No Free Lunch** direction.

This theorem is the crown jewel of classical statistical learning theory: it gives a single,
combinatorial quantity ($\text{VCdim}$) that completely characterizes learnability in the binary
classification setting.

## 🐍 Implementation from Scratch

Two implementations:

1. **Empirical VC dimension estimation** for linear classifiers: for a given dimension $d$,
   draw random point sets of increasing size $m$ and test whether *all* $2^m$ labelings can
   be realized by some linear classifier (checked via the perceptron algorithm with random
   restarts). The estimated VCdim is the largest $m$ where shattering succeeds.

2. **VC generalization bound computation:** given $d$, $n$, and $\delta$, compute the
   theoretical bound on $\sup_h |R(h) - \hat{R}_n(h)|$ and compare it to the empirical
   generalization gap of a linear classifier trained on synthetic data.

````python
# --- Part 1: Empirical VCdim estimation via brute-force shattering check ---

def can_linear_separate(X, labels):
    """Check if points X can be linearly separated to produce `labels`.
    Uses least-squares heuristic + perceptron with random restarts."""
    m, d = X.shape
    X_aug = np.hstack([X, np.ones((m, 1))])
    y = 2 * np.array(labels, dtype=float) - 1
    # Try least-squares first
    w, _, _, _ = np.linalg.lstsq(X_aug * y[:, None], np.ones(m), rcond=None)
    if np.all((X_aug @ w) * y > 1e-10):
        return True
    # Fallback: perceptron with random restarts
    rng_local = np.random.default_rng(42)
    for _ in range(50):
        w = rng_local.standard_normal(d + 1) * 0.1
        for _ in range(500):
            preds = X_aug @ w
            mistakes = (preds * y) <= 0
            if not np.any(mistakes):
                return True
            idx = np.where(mistakes)[0][0]
            w += y[idx] * X_aug[idx]
    return False

def check_shattering(X):
    """Check if linear classifiers shatter the point set X (all 2^m labelings)."""
    m = X.shape[0]
    for labeling in product([0, 1], repeat=m):
        if not can_linear_separate(X, labeling):
            return False
    return True

def estimate_vcdim_linear(d, n_trials=10, rng=None):
    """Estimate VCdim of linear classifiers in R^d by random shattering search."""
    if rng is None:
        rng = np.random.default_rng(0)
    max_shattered = 0
    for m in range(1, d + 4):
        if m > 8:  # cap at 2^8 = 256 labelings
            break
        found = False
        for _ in range(n_trials):
            X = rng.standard_normal((m, d))
            if check_shattering(X):
                found = True
                break
        if found:
            max_shattered = m
        else:
            break
    return max_shattered

print("Empirical VCdim estimation for linear classifiers in R^d:")
print(f"{'d':>4} {'Theory (d+1)':>14} {'Estimated':>10}")
print("-" * 32)
for d in [1, 2, 3, 4, 5]:
    est = estimate_vcdim_linear(d, n_trials=15, rng=np.random.default_rng(42))
    print(f"{d:4d} {d+1:14d} {est:10d}")

# --- Part 2: VC generalization bound vs empirical gap ---

def vc_bound(d, n, delta=0.05):
    """VC generalization bound: sqrt(8/n * (d*ln(2en/d) + ln(4/delta)))."""
    if d == 0:
        return np.sqrt(np.log(4 / delta) * 8 / n)
    return np.sqrt(8 / n * (d * np.log(2 * np.e * n / d) + np.log(4 / delta)))

print("\n--- VC bound vs empirical generalization gap ---")
rng = np.random.default_rng(0)
d_input = 5
w_true = rng.standard_normal(d_input + 1)

for n in [100, 500, 2000, 10000]:
    X_train = rng.standard_normal((n, d_input))
    X_train_aug = np.hstack([X_train, np.ones((n, 1))])
    y_train = (X_train_aug @ w_true >= 0).astype(float)
    flip = rng.random(n) < 0.05
    y_train[flip] = 1 - y_train[flip]
    w_hat, _, _, _ = np.linalg.lstsq(X_train_aug, y_train, rcond=None)
    train_err = np.mean((X_train_aug @ w_hat >= 0.5).astype(float) != y_train)
    X_test = rng.standard_normal((50000, d_input))
    X_test_aug = np.hstack([X_test, np.ones((50000, 1))])
    y_test = (X_test_aug @ w_true >= 0).astype(float)
    test_err = np.mean((X_test_aug @ w_hat >= 0.5).astype(float) != y_test)
    emp_gap = abs(test_err - train_err)
    bound = vc_bound(d_input + 1, n)
    print(f"n={n:5d} | train_err={train_err:.4f} | test_err={test_err:.4f} "
          f"| gap={emp_gap:.4f} | VC bound={bound:.4f} | holds: {emp_gap <= bound}")
````

## 🤖 Why This Matters for AI

The VC dimension was the first rigorous tool for understanding generalization, and its
limitations drive the modern theory of deep learning.

**Neural network VC dimension.** For a network with $W$ total weights and $L$ layers (using
piecewise-linear activations like ReLU), a classical result gives
$\text{VCdim} = O(WL \cdot \log W)$. For GPT-2 ($W \approx 1.5 \times 10^8$, $L = 12$),
this gives $\text{VCdim} \approx O(10^{10})$. Plugging into the VC bound with a realistic
training set ($n \approx 10^9$ tokens): the bound exceeds 1, meaning it guarantees **nothing**
— it's *vacuous*. Yet GPT-2 generalizes well empirically.

This gap between VC theory and practice is not a failure of theory — it's a signal that VC
dimension, which measures the *worst-case* complexity of the entire hypothesis class, is too
coarse for modern deep learning. The true generalization behavior depends on *which* functions
in $\mathcal{H}$ the training algorithm actually finds, not on the full class's expressive
capacity. This motivates:

- **Rademacher complexity** (`13.06`): a data-dependent complexity measure.
- **PAC-Bayes bounds** (recall `09.04`): bounds depending on distance from a prior.
- **Algorithmic stability** and **compression-based bounds**: accounting for SGD's
  inductive bias and overparameterized networks' implicit regularization.

Below, we compute the (vacuous) VC bound for networks of increasing size, demonstrating
concretely why VC theory alone cannot explain deep learning generalization.

````python
# --- VC bounds for neural networks: demonstrating vacuousness ---

def nn_vc_dim(W, L):
    """Approximate VC dimension of a neural net: O(W * L * log2(W))."""
    return W * L * np.log2(max(W, 2))

def vc_gen_bound(vcdim, n, delta=0.05):
    """VC generalization bound value."""
    if vcdim == 0 or n == 0:
        return float('inf')
    return np.sqrt(8 / n * (vcdim * np.log(2 * np.e * n / vcdim) + np.log(4 / delta)))

architectures = [
    ("Logistic regression (d=100)",     100,    1,   10000),
    ("Small MLP (2 hidden layers)",     50000,  3,   50000),
    ("ResNet-18",                       1.1e7,  18,  1.2e6),
    ("GPT-2 (124M)",                    1.24e8, 12,  1e9),
    ("GPT-3 (175B)",                    1.75e11, 96, 3e11),
]

print(f"{'Architecture':<35} {'Params':>12} {'VCdim':>14} {'n_train':>12} "
      f"{'VC bound':>12} {'Vacuous?':>10}")
print("-" * 100)
for name, W, L, n in architectures:
    vcd = nn_vc_dim(W, L)
    bound = vc_gen_bound(vcd, n)
    vacuous = bound > 1.0
    print(f"{name:<35} {W:>12.2e} {vcd:>14.2e} {n:>12.2e} "
          f"{bound:>12.4f} {'YES' if vacuous else 'no':>10}")

# --- Visualization: VC bound convergence rate ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

n_range = np.logspace(2, 7, 200)
for d_val, color, label in [(5, BLUE, 'd=5 (linear)'),
                             (50, GREEN, 'd=50 (small net)'),
                             (500, RED, 'd=500 (medium net)'),
                             (5000, PURPLE, 'd=5000 (larger net)')]:
    bounds = [vc_bound(d_val, n) if n > d_val else np.nan for n in n_range]
    ax1.loglog(n_range, bounds, color=color, lw=2, label=label)
ax1.axhline(y=1.0, color='gray', ls='--', alpha=0.5, label='vacuous (bound > 1)')
ax1.set_xlabel('Training set size $n$')
ax1.set_ylabel('VC generalization bound')
ax1.set_title('VC bound convergence rate\nfor different hypothesis complexities')
ax1.legend(fontsize=9); ax1.set_ylim(0.01, 100)

d_range = np.arange(1, 1001)
sample_complexity = [max(int(80 * d_val / 0.1**2), d_val + 1) for d_val in d_range]
ax2.loglog(d_range, sample_complexity, color=BLUE, lw=2)
ax2.set_xlabel('VC dimension $d$')
ax2.set_ylabel('Approx. samples for $\\varepsilon=0.1$')
ax2.set_title('Sample complexity scales linearly in $d$\n'
              + '(for fixed $\\varepsilon=0.1$, $\\delta=0.05$)')
ax2.fill_between(d_range, sample_complexity, alpha=0.15, color=BLUE)

plt.tight_layout()
plt.show()

print("\nKey insight: VC bounds are tight for simple models (logistic regression),")
print("but become vacuous for deep networks \u2014 motivating Rademacher complexity (13.06)")
print("and PAC-Bayes approaches that account for which subset of H the algorithm explores.")
````

1. A hypothesis class of linear classifiers in $\mathbb{R}^{999999}$ has
   $\text{VCdim} = 10^6$. Compute the VC generalization bound for $n = 10^7$ training
   examples with confidence $\delta = 0.05$. Is the bound vacuous (greater than 1)?

<details>
<summary>💡 Solution</summary>

Using $d = 10^6$, $n = 10^7$, $\delta = 0.05$:

$$\varepsilon = \sqrt{\frac{8}{10^7}\left(10^6 \ln\frac{2e \cdot 10^7}{10^6} + \ln\frac{4}{0.05}\right)}$$

$= \sqrt{\frac{8}{10^7}\left(10^6 \cdot \ln(2e \cdot 10) + \ln 80\right)}$
$= \sqrt{\frac{8}{10^7}\left(10^6 \cdot 3.688 + 4.382\right)}$
$= \sqrt{\frac{8 \cdot 3.688 \times 10^6}{10^7}} = \sqrt{2.950} \approx 1.718$.

The bound is **vacuous** ($> 1$). Even with $10\times$ more data than parameters, the VC
bound provides no useful guarantee — illustrating the $O(\sqrt{d/n})$ scaling.

</details>

### 💭 UNDERSTAND
2. The sinusoidal classifier $h_\omega(x) = \mathbf{1}[\sin(\omega x) \ge 0]$ has
   $\text{VCdim} = \infty$ despite being parameterized by a single real number $\omega$.
   Explain why the number of parameters is not a reliable proxy for VC dimension. Give
   an example of a hypothesis class with *many* parameters but *low* VC dimension, and one
   with *few* parameters but *high* VC dimension.

<details>
<summary>💡 Solution</summary>

VC dimension measures the *combinatorial richness* of the labeling patterns a class can
produce, not the dimensionality of its parameterization. A single continuous parameter
$\omega$ can encode arbitrarily complex binary patterns through the oscillatory structure of
$\sin(\omega x)$ — as $\omega \to \infty$, the sign changes become arbitrarily dense, allowing
any finite labeling to be achieved.

**Many parameters, low VCdim:** $\mathcal{H} = \{h_\theta : h_\theta(x) =
\mathbf{1}[\theta_1 + \theta_2 + \cdots + \theta_{1000} \ge 0]\}$ — 1000 parameters that
collapse to a single threshold, $\text{VCdim} = 0$.

**Few parameters, high VCdim:** The sinusoidal class (1 parameter, infinite VCdim).

The lesson: parameter counting is at best a rough heuristic. VC dimension is the correct
combinatorial measure of hypothesis class complexity.

</details>

3. Prove that $\text{VCdim}(\mathcal{H}) = d + 1$ for the class of affine linear
   classifiers $h_{w,b}(x) = \mathbf{1}[w^\top x + b \ge 0]$ on $\mathbb{R}^d$.
   *Hint:* The Theory section proves this for homogeneous classifiers on $\mathbb{R}^{d+1}$;
   explain why the affine case on $\mathbb{R}^d$ is equivalent (recall `01.03`).

<details>
<summary>💡 Solution</summary>

The affine classifier $h_{w,b}(x) = \mathbf{1}[w^\top x + b \ge 0]$ on $\mathbb{R}^d$ is
equivalent to the homogeneous classifier $h_{\tilde{w}}(\tilde{x}) =
\mathbf{1}[\tilde{w}^\top \tilde{x} \ge 0]$ on $\mathbb{R}^{d+1}$, via the standard
"lifting" map $x \mapsto \tilde{x} = (x, 1)$ and $\tilde{w} = (w, b)$. This is a linear
transformation (recall `01.03`) that preserves shattering: any set $\{x_1,\ldots,x_m\}$ in
$\mathbb{R}^d$ is shattered by affine classifiers iff $\{\tilde{x}_1,\ldots,\tilde{x}_m\}$
is shattered by homogeneous classifiers in $\mathbb{R}^{d+1}$.

Since the Theory section established $\text{VCdim} = (d+1)$ for homogeneous classifiers on
$\mathbb{R}^{d+1}$ (basis vectors for the lower bound, linear dependence for the upper), the
same value holds for affine classifiers on $\mathbb{R}^d$. $\blacksquare$

</details>

### 🐍 IMPLEMENT
4. Write a brute-force VC dimension estimator: given a classifier class (as a function
   that, given points $X$ and parameters, produces labels), systematically test shattering
   for $m = 1, 2, 3, \ldots$ until no shatterable set of size $m$ is found after $k$ random
   trials. Apply it to (a) threshold classifiers on $\mathbb{R}$ (should get 1), and (b)
   axis-aligned rectangles in $\mathbb{R}^2$ (should get 4).

<details>
<summary>💡 Solution</summary>

```python
def brute_force_vcdim(shatters_fn, dim, max_m=8, n_trials=20):
    rng = np.random.default_rng(0)
    max_shattered = 0
    for m in range(1, max_m + 1):
        found = False
        for _ in range(n_trials):
            X = rng.standard_normal((m, dim))
            if shatters_fn(X):
                found = True; break
        if found:
            max_shattered = m
        else:
            break
    return max_shattered
```

For threshold classifiers: `shatters(X)` sorts 1D points and checks all $2^m$ labelings,
verifying each can be achieved by some threshold. Returns VCdim = 1.

For axis-aligned rectangles in $\mathbb{R}^2$: $h_{a,b,c,d}(x) = \mathbf{1}[a \le x_1 \le b
\;\text{and}\; c \le x_2 \le d]$. Shattering check via exhaustive interval search.
Returns VCdim = 4.

</details>

### 🚀 CHALLENGE
6. Prove the Sauer–Shelah lemma by induction on $m + d$: if $\text{VCdim}(\mathcal{H}) = d$,
   then $\Pi_{\mathcal{H}}(m) \le \sum_{i=0}^{d} \binom{m}{i}$ for all $m$.
   *Hint:* Follow the proof sketch in the Theory section. The key step is arguing that the
   "both-values" subset $\mathcal{H}^\pm$ has VC dimension at most $d - 1$.

<details>
<summary>💡 Solution</summary>

**Proof by induction on $m + d$.**

Let $\Phi(m, d) = \sum_{i=0}^{d}\binom{m}{i}$.

*Base cases:*
- $d = 0$: $\mathcal{H}$ shatters no point, so $|\mathcal{H}|_C| \le 1 = \Phi(m, 0)$. ✓
- $m \le d$: $\Phi(m, d) \ge 2^m \ge |\mathcal{H}|_C|$. ✓

*Inductive step ($m > d \ge 1$):* Fix $x_1 \in C$, $C' = C \setminus \{x_1\}$.

Define $\mathcal{H}_0$ = projection of $\mathcal{H}|_C$ onto $C'$ (dropping the $x_1$
coordinate). Define $\mathcal{H}^\pm$ = labelings on $C'$ appearing with *both* values of
$h(x_1)$.

Counting: $|\mathcal{H}|_C| = |\mathcal{H}_0| + |\mathcal{H}^\pm|$.

$\mathcal{H}_0$ has VCdim $\le d$ on $m-1$ points: $|\mathcal{H}_0| \le \Phi(m-1, d)$.

$\mathcal{H}^\pm$ has VCdim $\le d-1$: if it shattered $S \subseteq C'$ with $|S| = d$,
then $\mathcal{H}$ would shatter $S \cup \{x_1\}$ (size $d+1$), contradicting VCdim = $d$.
So: $|\mathcal{H}^\pm| \le \Phi(m-1, d-1)$.

By Pascal's identity: $|\mathcal{H}|_C| \le \Phi(m-1, d) + \Phi(m-1, d-1) = \Phi(m, d)$.
$\blacksquare$

</details>
